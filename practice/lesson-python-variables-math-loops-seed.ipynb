## Setup

Run this once per session, after Check says the buttons are ready. Then run a lesson below.


```python
# Setup
# Run once per session, and again after practice_bank.py changes.
# Looks for practice_bank.py and questions.json next to this notebook,
# in notebooks/, or in /content (Colab's file panel).

import importlib.util
import sys
from pathlib import Path


def _find(name):
    folders = [
        Path.cwd(),
        Path.cwd() / "notebooks",
        Path("/content"),
        Path("/content/notebooks"),
    ]
    for folder in folders:
        candidate = folder / name
        if candidate.is_file():
            return candidate
    return None


bank_py = _find("practice_bank.py")
if bank_py is None:
    print("Could not find practice_bank.py.")
    print("Put practice_bank.py and questions.json in the same folder as this notebook.")
    print("In Colab: click the folder icon and upload both files.")
else:
    spec = importlib.util.spec_from_file_location("practice_bank", bank_py)
    practice_bank = importlib.util.module_from_spec(spec)
    sys.modules["practice_bank"] = practice_bank
    spec.loader.exec_module(practice_bank)
    try:
        problems = practice_bank.check_bank()
    except practice_bank.BankError as err:
        print(err)
    else:
        if problems:
            print("Fix questions.json before a lesson:")
            for line in problems:
                print("-", line)
        else:
            print("Setup ready.", practice_bank.bank_summary())
            print("Run a lesson cell, then click an answer.")
# Expected output:
# Setup ready. Then the category count and example count.
# Run a lesson cell, then click an answer.
```
